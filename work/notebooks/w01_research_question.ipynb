## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Decision:** given a content team's limited weekly refresh capacity, which pages should
be reviewed first — refreshed, expanded, protected, pruned, or monitored?

**Unit of analysis:** one page (one `content_hash_id`).

**Action:** a content manager pulls the top N pages from a ranked refresh queue —
each with a reason code attached — and assigns them to writers this week, instead of
picking pages by recency or gut feel.

**Cost of a wrong call:** the two error types cost differently. A false positive
(flagging a page that wasn't actually declining) wastes writer hours on a page that
didn't need it. A false negative (missing a page that IS declining) means lost
visibility keeps compounding unnoticed — likely the more expensive mistake over time.
The starter baseline can't weigh this asymmetry; a ranked, probability-based model can
